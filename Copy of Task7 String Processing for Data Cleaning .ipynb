```python

def clean_text(text):
    text = text.strip()
    text = text.lower()
    text = text.replace("-", " ")
    return text

name = "  NAGA-DURGA  "
email = "  DURGA@GMAIL.COM  "
category = "  DATA-SCIENCE  "

print("Cleaned Name:", clean_text(name))
print("Cleaned Email:", clean_text(email))
print("Cleaned Category:", clean_text(category))
```

```python

text = "  Python, Java, Data Science  "

print("Original:", text)
print("Strip:", text.strip())
print("Lower:", text.lower())
print("Upper:", text.upper())
print("Replace:", text.replace(",", "-"))
print("Split:", text.split(","))
```
